## Figuras: ganador (sin puentes) contra el mejor con un puente

```sage
for nm, c in certificados.items():
    print(nm)
    graphics_array([c["G"].dibujar(), c["H"].dibujar()]).show(figsize=[8, 3])
```

## Comparacion numerica de S_G - S_H

Solo como ilustracion: la prueba es q(1) > 0.

```sage
R = PolynomialRing(QQ, "p")
for nm, c in certificados.items():
    D = c["G"].polinomio_separabilidad(R) - c["H"].polinomio_separabilidad(R)
    print(nm, [(x, N(D(QQ(x)), digits=4)) for x in ["9/10", "99/100", "999/1000"]])
```
