```python
resultados = []
for n in range(7, 11):
    m_min = n + binomial(n - 2, 2)
    m_max = binomial(n, 2)
    for m in range(m_min, m_max + 1):
        G_globo = ConstruirGrafo2TerminalesBalloon(n, m)
        T2_globo = NumeroDe2Bosques(G_globo)
        T2_max, optimos = GrafosQueMaximizanT2(n, m)
        resultados.append((n, m, T2_globo, T2_max, len(optimos)))
        if T2_max > T2_globo:
            res_bool = True
        else: 
            res_bool = False
        print(f"n={n}, m={m}: #óptimos={len(optimos)}, contraejemplo?={res_bool}")
        if T2_max > T2_globo:
            candidato = optimos[0]
            print(f"  >>> CONTRAEJEMPLO: {sorted(candidato.grafo.edges(labels=False))}, s={candidato.s}, t={candidato.t}")
```

```python
contraejemplos = [r for r in resultados if r[3] > r[2]]
print(f"Pares (n,m) evaluados: {len(resultados)}")
print(f"Contraejemplos (algún grafo supera al globo en T2): {len(contraejemplos)}")
```
