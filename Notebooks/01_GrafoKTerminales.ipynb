## Arboles: F_{k-1} es un menor del Laplaciano

Subdivision equilibrada de K_{1,3} (k = 3) y doble estrella (k = 4).

```python
# K_{1,3} con patas (2, 2, 2): centro 0
T3 = Graph([(0, 1), (1, 2), (0, 3), (3, 4), (0, 5), (5, 6)])
A = GrafoKTerminales(T3, [2, 4, 6])
print("k = 3:", A.F(2), A.menor_laplaciano_terminales())

# doble estrella con patas 1 y camino central de largo 1: 0 - 1
T4 = Graph([(0, 1), (0, 2), (0, 3), (1, 4), (1, 5)])
B = GrafoKTerminales(T4, [2, 3, 4, 5])
print("k = 4:", B.F(3), B.menor_laplaciano_terminales())
```

## Isomorfismo con terminales

Los terminales se pueden permutar entre si, pero no con los demas vertices.

```python
K = graphs.CycleGraph(5)
print(GrafoKTerminales(K, [0, 1, 3]).es_isomorfo(GrafoKTerminales(K, [2, 3, 0])))   # True
print(GrafoKTerminales(K, [0, 1, 3]).es_isomorfo(GrafoKTerminales(K, [0, 1, 2])))   # False
```

## Control cruzado al azar

```python
import random
random.seed(int(1))
fallas = 0
pruebas = 0
for _ in range(200):
    n = random.randint(4, 7)
    Gr = graphs.RandomGNP(n, 0.5)
    if not Gr.is_connected() or Gr.size() > 14:
        continue
    k = random.randint(2, min(4, n))
    X = GrafoKTerminales(Gr, random.sample(list(Gr.vertices(sort=True)), k))
    pruebas += 1
    fallas += X.F_polinomio() != X.F_polinomio("aristas")
print(pruebas, "pruebas,", fallas, "fallas")
```
