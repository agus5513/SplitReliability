```python
def ImprimirGanador(n, m):
    try:
        G = ConstruirGrafo2TerminalesBalloon(n, m)
    except Exception:
        print("Valor de (n, m) no válido")
    T2G = NumeroDe2Bosques(G)
    b_valor, n_prima, lambda_prima = ParametrosEsqueletoBalloon(G)
    print(f"b = {b_valor}, n' = {n_prima}, λ' = {lambda_prima}, diferencia = {n_prima -lambda_prima}")
    for k in (0, 1, 2):
        try:
            H = VariantesBalloonH(k, n, m)
        except:
            T2H = -1
        T2H = NumeroDe2Bosques(H)
        diferencia = T2H - T2G
        ganador = "H" if diferencia > 0 else ("G" if diferencia < 0 else "empate")
        print(f"k={k}:(gana {ganador})")
    
```

```python
ImprimirGanador(9, 23)
```
