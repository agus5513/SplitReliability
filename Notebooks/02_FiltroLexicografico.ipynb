## Arboles optimos tau_k(L)

Para k = 4 la estrella solo gana en L = 4; despues gana la doble estrella.

```python
for k, Ls in [(3, range(3, 11)), (4, range(4, 13)), (5, range(5, 13))]:
    for L in Ls:
        valor, arboles = tau(k, L)
        grados = sorted({tuple(sorted(d for d in T.degree() if d >= 3)) for T in arboles})
        print(f"k = {k}, L = {L}: tau = {valor}, grados de ramificacion {grados}")
```
